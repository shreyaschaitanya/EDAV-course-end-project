## Course Outcome 1 (CO1) / UNIT - I: NumPy Basics
**Syllabus Topics Covered**:
- Creating `ndarrays`, Data Types for `ndarrays`
- Arithmetic with NumPy Arrays
- Basic Indexing, Slicing, Boolean Indexing, Fancy Indexing
- Transposing Arrays and Swapping Axes
- Pseudo Random Number Generation & Universal Functions (ufuncs)

```python
import numpy as np

# 1. Creating ndarrays and checking Data Types
sales_volume = np.array([4500.5, 5200.0, 4800.75, 6100.2, 3900.0], dtype=np.float64)
print("Sales Volume Array (Liters):", sales_volume)
print("Data Type:", sales_volume.dtype)

# 2. Arithmetic with NumPy Arrays (Price = 100 INR/Liter)
fuel_price = 100.0
revenue_array = sales_volume * fuel_price
print("Calculated Revenue (INR):", revenue_array)

# 3. Basic Indexing, Slicing, Boolean Indexing, Fancy Indexing
print("First 3 sales:", sales_volume[:3])
high_sales = sales_volume[sales_volume > 5000]
print("Boolean Indexing (Sales > 5000L):", high_sales)
fancy_indexed = sales_volume[[0, 3, 4]]
print("Fancy Indexing (Indices 0, 3, 4):", fancy_indexed)

# 4. Transposing Arrays and Swapping Axes
matrix_data = np.array([[4500, 100], [5200, 102], [4800, 101]]) # [Volume, Price]
print("Original Shape:", matrix_data.shape)
transposed_matrix = matrix_data.T
print("Transposed Matrix Shape:", transposed_matrix.shape)

# 5. Pseudo Random Number Generation & Universal Functions (ufuncs)
np.random.seed(42)
simulated_demand = np.random.normal(loc=5000, scale=500, size=10)
print("Simulated Daily Demand (Random Normal):", np.round(simulated_demand, 2))
print("Universal Function (Square Root):", np.sqrt(simulated_demand))
```

## Course Outcome 2 (CO2) / UNIT - II: Getting Started with Pandas & Data I/O
**Syllabus Topics Covered**:
- Series, DataFrame, Index Objects
- Reindexing, Dropping Entries, Indexing, Selection and Filtering
- Arithmetic & Data Alignment, Function Application and Mapping
- Sorting and Ranking, Duplicate Labels, Descriptive Statistics
- Data Loading: Reading Text Files in Pieces (Chunking), Writing Data

```python
import pandas as pd

# Generating synthetic dataset (>10,000 rows & multi-attribute features)
np.random.seed(42)
num_rows = 10000

dates = pd.date_range(start="2023-01-01", periods=num_rows, freq="h")
station_ids = np.random.choice(["ST_001", "ST_002", "ST_003", "ST_004"], size=num_rows)
fuel_types = np.random.choice(["Regular Petrol", "Premium Petrol", "Diesel", "CNG"], size=num_rows)
quantity_liters = np.random.uniform(5, 70, size=num_rows)
price_per_liter = np.random.choice([96.5, 104.2, 88.0, 75.5], size=num_rows)
payment_methods = np.random.choice(["UPI", "Credit Card", "Cash", "Debit Card"], size=num_rows)

df = pd.DataFrame({
    'Timestamp': dates,
    'Station_ID': station_ids,
    'Fuel_Type': fuel_types,
    'Quantity_Liters': np.round(quantity_liters, 2),
    'Price_Per_Liter': price_per_liter,
    'Payment_Method': payment_methods
})
df['Total_Amount'] = df['Quantity_Liters'] * df['Price_Per_Liter']

# Save to File & Read in Pieces (Chunking)
df.to_csv("petrol_station_sales.csv", index=False)
print("Saved dataset with 10,000 rows to 'petrol_station_sales.csv'.")

# Reading Text Files in Pieces
chunk_list = []
for chunk in pd.read_csv("petrol_station_sales.csv", chunksize=2500):
    chunk_list.append(chunk)

df_loaded = pd.concat(chunk_list, axis=0)

# DataFrame Summarizing & Descriptive Statistics
print("\nDescriptive Statistics:")
print(df_loaded[['Quantity_Liters', 'Total_Amount']].describe())

# Selection, Filtering, Sorting and Function Mapping
high_val = df_loaded[df_loaded['Total_Amount'] > 4000].sort_values(by='Total_Amount', ascending=False)
print("\nTop 5 High-Value Transactions:")
print(high_val.head())

df_loaded['Sales_Category'] = df_loaded['Total_Amount'].apply(lambda x: 'High' if x > 3500 else 'Medium' if x > 1500 else 'Low')
print("\nSales Category Breakdown:\n", df_loaded['Sales_Category'].value_counts())
```

## Course Outcome 3 (CO3) / UNIT - III: Data Cleaning, Preparation & Wrangling
**Syllabus Topics Covered**:
- Filtering out & Filling Missing Data, Transforming Data using Function/Mapping
- Replacing Values, Renaming Axis Indexes, Discretization & Binning
- Categorical Extension Type in Pandas & Computations
- Hierarchical Indexing (Reordering, Sorting, Summary Stats)
- Database-Style DataFrame Joins, Merging on Index, Concatenation

```python
# Introduce NaN values to demonstrate cleaning
df_clean = df_loaded.copy()
df_clean.loc[100:150, 'Quantity_Liters'] = np.nan

# 1. Filling Missing Data
print("Missing Quantity Count:", df_clean['Quantity_Liters'].isna().sum())
df_clean['Quantity_Liters'] = df_clean['Quantity_Liters'].fillna(df_clean['Quantity_Liters'].median())

# 2. Discretization and Binning & Categoricals
df_clean['Tank_Size_Bin'] = pd.cut(df_clean['Quantity_Liters'], bins=[0, 15, 40, 100], labels=['Small', 'Medium', 'Large'])
print("\nTank Size Bin Counts:\n", df_clean['Tank_Size_Bin'].value_counts())

# 3. Database-Style Joins & Merging
station_metadata = pd.DataFrame({
    'Station_ID': ["ST_001", "ST_002", "ST_003", "ST_004"],
    'City': ["Hyderabad", "Bengaluru", "Chennai", "Mumbai"],
    'Storage_Capacity_KL': [50, 75, 60, 100]
})

merged_df = pd.merge(df_clean, station_metadata, on='Station_ID', how='inner')
print("\nMerged DataFrame Sample:")
print(merged_df[['Station_ID', 'City', 'Fuel_Type', 'Total_Amount']].head())

# 4. Hierarchical Indexing
hierarchical_df = merged_df.set_index(['City', 'Fuel_Type']).sort_index()
print("\nHierarchical Index Sample (Hyderabad):")
print(hierarchical_df.loc['Hyderabad'].head())
```

## Course Outcome 4 (CO4) / UNIT - IV: Data Aggregation & Time Series Analysis
**Syllabus Topics Covered**:
- Column-Wise and Multiple Function Application (Split-Apply-Combine)
- Returning Aggregated Data without Row Indexes
- Time Series: Converting String to Datetime, Indexing, Selection, Subsetting
- Generating Date Ranges, Frequencies & DateOffsets
- Resampling & Frequency Conversion (Downsampling, Upsampling, Interpolation)

```python
# 1. Data Aggregation & Split-Apply-Combine
agg_summary = merged_df.groupby(['City', 'Fuel_Type'], as_index=False).agg(
    Total_Liters=('Quantity_Liters', 'sum'),
    Average_Revenue=('Total_Amount', 'mean'),
    Max_Transaction=('Total_Amount', 'max'),
    Txn_Count=('Total_Amount', 'count')
)
print("Aggregated Sales Summary (Without Row Index):")
print(agg_summary)

# 2. Time Series Analysis
merged_df['Timestamp'] = pd.to_datetime(merged_df['Timestamp'])
ts_df = merged_df.set_index('Timestamp')

# Downsampling to Daily Total Sales
daily_sales = ts_df['Total_Amount'].resample('D').sum()
print("\nDaily Resampled Revenue (First 5 Days):\n", daily_sales.head())

# Upsampling and Linear Interpolation
upsampled = daily_sales.head(5).resample('6H').interpolate(method='linear')
print("\nUpsampled (6-Hour Interval) Revenue Sample:\n", upsampled.head(8))
```

## Course Outcome 5 (CO5) / UNIT - V: Plotting and Visualization
**Syllabus Topics Covered**:
- Figures and Subplots, Colors, Markers, Line Styles, Ticks, Labels, Legends
- Saving Plots to File
- Seaborn: Line Plots, Bar Plots, Histograms/Density Plots, Scatter Plots, FacetGrids

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

# 1. Matplotlib Figure & Subplots
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Subplot 1: Daily Revenue Trend (Line Plot with markers and custom line style)
daily_sales.head(30).plot(ax=axes[0], color='teal', marker='o', linestyle='--', linewidth=1.5)
axes[0].set_title('30-Day Petrol Station Revenue Trend (CO5 Line Plot)', fontsize=12)
axes[0].set_xlabel('Date')
axes[0].set_ylabel('Total Revenue (INR)')

# Subplot 2: Seaborn Bar Plot (Revenue per City)
sns.barplot(data=merged_df, x='City', y='Total_Amount', hue='Fuel_Type', ax=axes[1], palette='Set2')
axes[1].set_title('Average Transaction Amount by City & Fuel Type', fontsize=12)
axes[1].set_xlabel('City')
axes[1].set_ylabel('Average Amount (INR)')

plt.tight_layout()
plt.savefig('daily_revenue_and_city_bar.png', dpi=300)
plt.show()

# 2. Seaborn Histogram & Density Plot (KDE)
plt.figure(figsize=(10, 5))
sns.histplot(data=merged_df, x='Quantity_Liters', hue='Fuel_Type', kde=True, palette='magma', element="step")
plt.title('Fuel Demand Quantity Density & Histogram Distribution', fontsize=14)
plt.xlabel('Fuel Quantity (Liters)')
plt.ylabel('Density Count')
plt.savefig('fuel_demand_histogram.png', dpi=300)
plt.show()

# 3. Seaborn FacetGrid & Scatter Plot
g = sns.FacetGrid(merged_df, col="City", hue="Fuel_Type", palette="tab10", height=4, aspect=1.2)
g.map(sns.scatterplot, "Quantity_Liters", "Total_Amount", alpha=0.5)
g.add_legend()
g.fig.suptitle('Quantity vs Revenue Facet Grid Across Cities', y=1.03, fontsize=14)
plt.savefig('facet_grid_scatter.png', dpi=300)
plt.show()

print("All visualizations rendered and saved successfully.")
```
